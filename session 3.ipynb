```python
import numpy as np

# ---------------------------------------------------------
# Task 1: Two friends' daily steps over a week (element-wise ops)
# ---------------------------------------------------------
friend1 = np.array([8000, 9500, 7200, 10000, 8800, 12000, 6500])
friend2 = np.array([7500, 10200, 8000, 9000, 9100, 11000, 7000])

print("Friend 1 steps:", friend1)
print("Friend 2 steps:", friend2)
print("Addition (combined steps):", friend1 + friend2)
print("Subtraction (F1 - F2):    ", friend1 - friend2)
print("Multiplication:           ", friend1 * friend2)
print("Division (F1 / F2):       ", np.round(friend1 / friend2, 2))
print()

```

```python
import numpy as np

user_preferences = np.array([
    [5, 3, 0],
    [4, 0, 2],
    [1, 5, 4],
])
song_popularity = np.array([
    [2, 1, 3],
    [0, 4, 1],
    [5, 2, 0],
])

rec_dot = np.dot(user_preferences, song_popularity)
rec_matmul = np.matmul(user_preferences, song_popularity)

print("Recommendation matrix using np.dot():")
print(rec_dot)
print("Recommendation matrix using np.matmul():")
print(rec_matmul)
print("Are both results equal?", np.array_equal(rec_dot, rec_matmul))

# Difference between dot() and matmul():
# For 2D matrices (like here) both perform standard matrix multiplication and
# give the same result. They differ for other cases:
#  - Scalars: np.dot(2, A) works, but np.matmul does not allow scalar operands.
#  - N-D arrays (3D+): np.matmul treats them as stacks of matrices and
#    multiplies matrix-by-matrix over the last two dimensions (with
#    broadcasting), while np.dot does a sum-product over the last axis of the
#    first array and the second-to-last axis of the second, giving a
#    different, larger output shape.
print()

```

```python
import numpy as np


image = np.array([
    [ 10,  50, 100, 150],
    [ 20,  60, 110, 160],
    [ 30,  70, 120, 170],
    [ 40,  80, 130, 180],
])

rotated = image.T
print("Original image:")
print(image)
print("Transposed (rotated) image:")
print(rotated)
print("Mean:              ", np.mean(rotated))
print("Median:            ", np.median(rotated))
print("Standard deviation:", np.std(rotated))
print("Variance:          ", np.var(rotated))
print()

```

```python
import numpy as np


orders = np.array([
    [12, 15, 14, 18, 20, 22],   # City 1 over 6 days
    [10, 13, 16, 17, 19, 21],   # City 2 over 6 days
])
print("Orders (2, 6):")
print(orders)

reshaped = orders.reshape(3, 4)
print("Reshaped to (3, 4):")
print(reshaped)

flat = reshaped.flatten()
print("Flattened:", flat)

part1, part2 = np.split(flat, 2)
print("Part 1:", part1)
print("Part 2:", part2)

stacked = np.vstack((part1, part2))
print("Stacked vertically:")
print(stacked)
```
